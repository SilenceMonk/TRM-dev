# TRM TinyStories D256
训练 loss 与固定验证子集的 loss/PPL；完整验证单独标记。

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
root = Path.cwd()
if root.name == 'trm_tinystories': root = root.parent
run = root / 'runs/trm_tinystories/d256_seed0'
train = [json.loads(x) for x in (run / 'train.jsonl').read_text().splitlines()]
val = [json.loads(x) for x in (run / 'validation.jsonl').read_text().splitlines()]
subset = [x for x in val if not x['full_validation']]
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot([x['step'] for x in train], [x['train_loss'] for x in train], label='train')
ax[0].plot([x['step'] for x in subset], [x['loss'] for x in subset], label='validation subset')
ax[0].set(xlabel='optimizer update', ylabel='token cross entropy'); ax[0].legend()
ax[1].plot([x['step'] for x in subset], [x['ppl'] for x in subset])
ax[1].set(xlabel='optimizer update', ylabel='validation subset PPL', yscale='log')
plt.tight_layout()
val[-1]
```
